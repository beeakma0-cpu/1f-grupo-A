faça um programa que receba um numero e imprime uma piramide perfeitamente centralizada

```python
#Programa que recebe um numero e imprime uma pirâmide perfeitamente centralizada

palavra = input("Digite um número: ")
numero = int(palavra)

for i in range(1, numero + 1):
    print(" " * (numero - i) + str(i) * (2 * i - 1))
```
