```python
# SISTEMA DE ELEIÇÃO

# Lista de candidatos
candidatos = {
    1: "Candidato 1",
    2: "Candidato 2",
    3: "Candidato 3"
}

# Quantidade de votos de cada candidato
votos = {
    1: 0,
    2: 0,
    3: 0
}

# Contadores
total_eleitores = 0
total_votaram = 0
votos_opcionais = 0
votos_obrigatorios = 0

print("=" * 40)
print("SISTEMA DE ELEIÇÃO")
print("=" * 40)

while True:

    print("-- CADASTRO DO ELEITOR --")

    nome = input("Digite o nome do eleitor: ")
    idade = int(input("Digite a idade do eleitor: "))

    total_eleitores += 1

    # Verificação da situação eleitoral

    if idade < 16:
        print(f"{nome} não pode votar.")
        print("Motivo: menor de 16 anos.")

    elif idade == 16 or idade == 17:
        print(f"{nome} pode votar.")
        print("Situação: voto opcional.")

        resposta = input("Deseja votar? (sim/não): ").lower()

        if resposta == "sim":
            print("-- CANDIDATOS --")

            for numero, nome_candidato in candidatos.items():
                print(f"{numero} - {nome_candidato}")

            while True:
                try:
                    numero = int(input("Digite o número do candidato: "))

                    if numero in candidatos:
                        votos[numero] += 1
                        total_votaram += 1
                        votos_opcionais += 1

                        print("Voto registrado com sucesso! :)")
                        break
                    else:
                        print("Candidato inválido! Digite um número válido.")

                except ValueError:
                    print("Digite apenas números.")

        else:
            print("Eleitor optou por não votar.")

    else: # idade >= 18
        print(f"{nome} deve votar.")
        print("Situação: voto obrigatório.")

        print("--- CANDIDATOS ---")

        for numero, nome_candidato in candidatos.items():
            print(f"{numero} - {nome_candidato}")

        while True:
            try:
                numero = int(input("Digite o número do candidato: "))

                if numero in candidatos:
                    votos[numero] += 1
                    total_votaram += 1
                    votos_obrigatorios += 1

                    print("Voto registrado com sucesso!")
                    break
                else:
                    print("Candidato inválido! Digite um número válido.")

            except ValueError:
                print("Digite apenas números.")

    # Pergunta se deseja cadastrar outro eleitor
    continuar = input("Deseja cadastrar outro eleitor? (sim/não): ").lower()

    if continuar == "não":
        break


print("=" * 40)
print(" RESULTADO DA ELEIÇÃO")
print("=" * 40)

print(f"Total de eleitores cadastrados: {total_eleitores}")
print(f"Total de eleitores que votaram: {total_votaram}")
print(f"Quantidade de votos opcionais: {votos_opcionais}")
print(f"Quantidade de votos obrigatórios: {votos_obrigatorios}")

print("--- VOTOS POR CANDIDATO ---")

for numero, nome_candidato in candidatos.items():
    print(f"{numero} - {nome_candidato}: {votos[numero]} voto(sim)")


# Encontrando a maior quantidade de votos
maior_quantidade = max(votos.values())

# Encontrando os candidatos que possuem a maior quantidade
vencedores = []

for numero, quantidade in votos.items():
    if quantidade == maior_quantidade:
        vencedores.append(numero)


# Verificação de vencedor ou empate
if len(vencedores) == 1:
    numero_vencedor = vencedores[0]
    print("--- VENCEDOR ---")
    print(f"Candidato vencedor: {candidatos[numero_vencedor]}")
    print(f"Número: {numero_vencedor}")
    print(f"Votos: {votos[numero_vencedor]}")

else:
    print("--- EMPATE ---")
    print("Houve empate entre os candidatos:")

    for numero in vencedores:
        print(
            f"- {candidatos[numero]} "
            f"(número {numero}): {votos[numero]} voto(s)"
        )

print("Fim da votação!")
```
