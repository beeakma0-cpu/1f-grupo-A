```python
#Calculadora com as 4 operações

#Menu
#Opção 1 - Somar
#Opção 2 - Subtrair
#Opção 3 - Multiplicar
#Opção 4 - Divisão
#Opção 5 - Sair

opcao = 0
while opcao != 5:
  print("---- Calculadora ----")
  print("")
  print("1 - Soma")
  print("")
  print("2 - Subtração")
  print("")
  print("3 - Multiplicação")
  print("")
  print("4 - Divisão")
  print("")
  print("5 - Sair")
  print("")

  try:
    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
      print("Você escolheu Soma.")

      num1 = float(input("Digite o primeiro número: "))
      num2 = float(input("Digite o segundo número: "))

      resultado = num1 + num2

      print(f"O resultado da soma é: {resultado}")


    elif opcao == 2:
      print("\nVocê escolheu Subtração.")

      num1 = float(input("Digite o primeiro número: "))
      num2 = float(input("Digite o segundo número: "))

      resultado = num1 - num2
      print(f"O resultado da subtração é: {resultado}")

    elif opcao == 3:
      print("\nVocê escolheu Multiplicação.")

      num1 = float(input("Digite o primeiro número: "))
      num2 = float(input("Digite o segundo número: "))

      resultado = num1 * num2
      print(f"O resultado da multiplicação é: {resultado}")

    elif opcao == 4:
      print("\nVocê escolheu Divisão.")

      num1 = float(input("Digite o primeiro número: "))
      num2 = float(input("Digite o segundo número: "))

      if num2 != 0:
        resultado = num1 / num2
        print(f"O resultado da divisão é: {resultado}")

      else:
        print("Erro: Não é possível dividir por zero!")

    elif opcao == 5:
      print("Saindo da calculadora.")

    else:
      print("Opção inválida. Por favor, escolha uma opção entre 1 e 5.")

  except ValueError:
    print("Entrada inválida. Por favor, digite um número válido.")

  except Exception as e:
    print(f"Ocorreu um erro inesperado: {e}")

print("Programa encerrado.")
```

```python
print('------------------------------------------------------------')
print("")
print('------------ Bem-vindo ao Jogo Inteligente! ----------------')
print("")
print('------------------------------------------------------------')
print("")

opcao = 0
while opcao != 4:
  print('---- Menu ----')
  print("")
  print('1 - Verificar se o número é Par')
  print("")
  print('2 - Verificar se o número é Positivo')
  print("")
  print('3 - Verificar se o número é Maior que 2')
  print("")
  print('4 - Sair')
  print("")

  try:
    opcao = int(input('Escolha uma opção: '))

    if opcao == 1:
      print('\nOpção 1: Verificar Par')

      num = int(input('Digite um número: '))

      if num % 2 == 0:
        print(f'O número {num} é PAR.')

      else:
        print(f'O número {num} é ÍMPAR.')

    elif opcao == 2:
      print('\nOpção 2: Verificar Positivo')

      num = int(input('Digite um número: '))

      if num > 0:
        print(f'O número {num} é POSITIVO.')

      elif num < 0:
        print(f'O número {num} é NEGATIVO.')

      else:
        print(f'O número {num} é ZERO.')

    elif opcao == 3:
      print('\nOpção 3: Verificar Maior que 2')

      num = int(input('Digite um número: '))

      if num > 2:
        print(f'O número {num} é MAIOR que 2.')

      else:
        print(f'O número {num} NÃO é maior que 2.')

    elif opcao == 4:
      print('Saindo do Jogo Inteligente. Até mais!')

    else:
      print('Opção inválida. Por favor, escolha uma opção entre 1 e 4.')

  except ValueError:
    print('Entrada inválida. Por favor, digite um número para a opção ou para a verificação.')

  except Exception as e:
    print(f'Ocorreu um erro inesperado: {e}')

print('Programa Jogo Inteligente encerrado.')
```

```python
#Programa que verifica a senha, 3 tentativas, caso erre, imprimir mensagem "acesso negado, tente novamente mais tarde!" e encerrar o programa

senha_correta = 44
tentativas = 0
max_tentativas = 3
acesso_concedido = False

while tentativas < max_tentativas:
  try:
    senha_digitada = int(input(f"Tentativa {tentativas + 1}/{max_tentativas}. Digite sua senha: "))

    if senha_digitada == senha_correta:
      print("Acesso concedido!")

      acesso_concedido = True
      break
    else:
      print("Senha incorreta.")
      tentativas += 1

  except ValueError:
    print("Entrada inválida. Por favor, digite apenas números.")

    tentativas += 1


if not acesso_concedido:
  print("Acesso negado, tente novamente mais tarde!")

print("Programa encerrado.")
```
