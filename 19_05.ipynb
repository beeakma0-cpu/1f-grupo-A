```python
#Programa que recebe as notas dos 04 bimestre e imprime a media com o resultado de aprovado, recuperação ou reprovado

#REGRAS
#Utilizar while
#Nota maior ou igual que 7 = aprovado
#Nota maior ou igual que 5 = recuperação
#Nota manor que 5 = reprovado

nota = 0
bimestre =1
resultado = 0

while bimestre <= 4:
  nota = float (input(f"digite a sua nota do {bimestre}º bimestre: "))
  bimestre += 1
  resultado = resultado + nota

if (resultado /4) >= 7:
  print(f"Sua media foi {resultado /4}, você foi APROVADO!!😉")

elif (resultado /4) >= 5:
    print(f"Sua media foi {resultado /4}, você esta de RECUPERAÇÃO!!🙂")

else:
    print(f"Sua media foi {resultado /4}, você está REPROVADO!!😶")
```

```python
#Programa caixa eletronico
#O programa deverá funcionar através de um menu interativo utilizando "while"
#O programa deve iniciar com um saldo inicial de R$1000,00 e permitir que o usuário escolha diferentes operações até decidir sair do sistema

saldo = 1000.00 # Corrected to float
opcao = 0 # Consistent variable name

while opcao != 4:
    print("\nCaixa eletrônico")
    print("1 - Consultar saldo")
    print("2 - Depositar")
    print("3 - Sacar")
    print("4 - Sair")

    try:
        opcao = int(input("Escolha uma opção: "))

        if opcao == 1:
            print(f"Seu saldo é de: R${saldo:.2f}") # Display actual saldo
        elif opcao == 2:
            try:
                deposito = float(input("Digite o valor do depósito: R$"))
                if deposito > 0:
                    saldo += deposito
                    print("Depósito realizado com sucesso!")
                else:
                    print("Valor inválido. O depósito deve ser maior que zero.")
            except ValueError:
                print("Entrada inválida para o depósito. Digite um número.")
        elif opcao == 3:
            try:
                saque = float(input("Digite o valor do saque: R$"))
                if saque > 0: # Ensure positive withdrawal amount
                    if saque <= saldo:
                        saldo -= saque
                        print("Saque realizado com sucesso!")
                    else:
                        print("Saldo insuficiente.")
                else:
                    print("Valor inválido. O saque deve ser maior que zero.")
            except ValueError:
                print("Entrada inválida para o saque. Digite um número.")
        elif opcao == 4:
            print("Saindo do caixa eletrônico. Obrigado!")
        else:
            print("Opção inválida. Tente novamente.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número correspondente à opção.")
```
